1. Import Libraries

```python
# Import required libraries
import pandas as pd
import numpy as np
```

2. Load Dataset

```python
# Load the dataset
df = pd.read_csv("AdventureWorks Product Subcategories Lookup.csv", encoding="ISO-8859-1")

# Preview dataset
df.head()
```

3. Before Dataset Overview

```python
# Check dataset structure
df.info()

# Check number of rows and columns
print(df.shape)

# Statistical summary
df.describe()
```

4. Check Duplicate Rows

```python
# Check duplicate 
df.duplicated().sum()
```

5. Check Missing Values

```python
# Check missing values
df.isnull().sum()
```

6. Fix Data Types

```python
# Clean and convert to category
df['SubcategoryName'] = df['SubcategoryName'].astype('category')
df.dtypes
```

7. Clean Text Columns

```python
# Standardize capitalization
df['SubcategoryName'] = df['SubcategoryName'].str.strip().str.title()

# Convert back to categorical type
df['SubcategoryName'] = df['SubcategoryName'].astype('category')

# Verify
print(df.dtypes)
print(df.head())
```

8. check unique categories

```python
# Count unique subcategories
print("No. of unique subcategories : " ,df['SubcategoryName'].nunique())

#To see the actual unique values
print(df['SubcategoryName'].unique())
```

9. After Data cleaning

```python
# 1️.Check Missing Values
print(df.isnull().sum())

# 2.Check Duplicate Rows
print(df.duplicated().sum())

# 3.Verify Final Dataset Shape
print(df.shape)

# 4.Datatype
print(df.dtypes)

# 5.Data preview
print(df.head(5))
```

10. Save Cleaned Dataset

```python
df.to_csv("Cleaned_Productsubcategory_Data.csv", index=False)
```
