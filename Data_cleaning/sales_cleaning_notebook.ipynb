# Data Cleaning Notebook

1. Import libraries

```python
# Import libraries
import pandas as pd
import numpy as np
```

2. Load Dataset

```python
# Load the dataset
df = pd.read_csv("AdventureWorks Sales Data 2020.csv", encoding="ISO-8859-1")

# Preview dataset
df.head()
```

3. Check Duplicate Rows

```python
# Check duplicate rows
df.duplicated().sum()
```

4. Check Missing Values

```python
# Check missing values
df.isnull().sum()
```

5. Fix Data Types

```python
# Step 1: Check original datatypes ---
print("Before conversion:")
print(df.info())

# Step 2: Convert datatypes ---
df['OrderDate'] = df['OrderDate'].astype('datetime64[s]')
df['StockDate'] = df['StockDate'].astype('datetime64[s]')

# Step 3: Check datatypes after conversion ---
print("\nAfter conversion:")
print(df.info())
```

6. Clean Text Columns

```python
# Remove leading and trailing spaces
df['OrderNumber'] = df['OrderNumber'].str.strip()

# Convert to uppercase for consistency
df['OrderNumber'] = df['OrderNumber'].str.upper()
df.head()
```

7. check unique categories

```python
# Check unique values in categorical columns
unique_summary = pd.DataFrame({
    'Column': df.columns,
    'Unique_Count': [df[col].nunique() for col in df.columns],
    'Sample_Values': [df[col].unique()[:5] for col in df.columns]  
# first 5 unique values as sample
})

# Display summary
print(unique_summary)
```

8. Remove unnecessary columns

```python
df = df.drop(columns=['OrderLineItem', 'OrderNumber'])
df.columns
```

9. Feature engineering

```python
# yearly / monthly sales analysis
df['OrderYear'] = df['OrderDate'].dt.year
df['OrderMonth'] = df['OrderDate'].dt.month

# visual dashboards
df['OrderMonthName'] = df['OrderDate'].dt.month_name()

# days between stock and order.
df['ProcessingDays'] = (df['OrderDate'] - df['StockDate']).dt.days
df.columns
```

10. Arrange columns

```python
df = df[['OrderDate',
         'StockDate',
         'OrderYear',
         'OrderMonth',
         'OrderMonthName',
         'ProductKey',
         'CustomerKey',
         'TerritoryKey',
         'OrderQuantity',
         'ProcessingDays']]

df.head()
```

11. Outlier detection

```python
Q1 = df['OrderQuantity'].quantile(0.25)
Q3 = df['OrderQuantity'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['OrderQuantity'] < lower_bound) | (df['OrderQuantity'] > upper_bound)]

print(outliers)
```

12. Final Data Validation

```python
# 1️.Check Missing Values
print(df.isnull().sum())

# 2.Check Duplicate Rows
print(df.duplicated().sum())

# 3.Reset Index After Row Removal
df = df.reset_index(drop=True)

# 4.Verify Final Dataset Shape
print(df.shape)

# 5.Data preview
print(df.head(5))

# 6.Save Cleaned Dataset
df.to_csv("Cleaned_Sales_Data.csv", index=False)
```
